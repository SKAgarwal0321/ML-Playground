NUMPY INDEXING AND SLICING OF MATIX

```python
b = np.arange(1,31).reshape(6,5)
b
```

```python
b[5]
```

```python
b[0,0]
```

```python
b[0:2,1:3]
```

```python
b[3:6,3:5]
```

```python
b[:,2]
```
