MATRIX OPRATORS

```python
A = np.arange(1,5).reshape(2,2)
B = np.arange(5,9).reshape(2,2)
```

```python
A
```

```python
B
```

```python
# Matrix multiplication
A@B
np.dot(A,B)
```

```python
#Transpose
A.T
```

```python
#Determinant of A
det_A = np.linalg.det(A).round(0)
det_A
```

```python
#inverse
np.linalg.inv(A)
```

```python
#Adjoint
adj_A = det_A * np.linalg.inv(A)
adj_A
```
